```python
import os, urllib.request, ssl

DESTINO = '/content/drive/MyDrive/Mecatronica/Fuentes citadas (papers y libros)'
os.makedirs(DESTINO, exist_ok=True)

FICHEROS = [
    ('https://arxiv.org/pdf/1512.03385', 'B6_S27_ResNet_He_2015.pdf'),
    ('https://arxiv.org/pdf/1506.02640', 'B6_S27_YOLO_Redmon_2016.pdf'),
    ('https://arxiv.org/pdf/1506.01497', 'B6_S27_FasterRCNN_Ren_2015.pdf'),
    ('https://arxiv.org/pdf/1411.4038', 'B6_S27_FCN_Long_2014.pdf'),
    ('https://arxiv.org/pdf/1505.04597', 'B6_S27_UNet_Ronneberger_2015.pdf'),
    ('https://arxiv.org/pdf/1703.06870', 'B6_S27_MaskRCNN_He_2017.pdf'),
    ('https://arxiv.org/pdf/2010.11929', 'B6_S27_ViT_Dosovitskiy_2020.pdf'),
    ('https://arxiv.org/pdf/2304.02643', 'B6_S27_SAM_Kirillov_2023.pdf'),
    ('https://proceedings.neurips.cc/paper_files/paper/2012/file/c399862d3b9d6b76c8436e924a68c45b-Paper.pdf', 'B6_S27_AlexNet_Krizhevsky_2012.pdf'),
    ('https://arxiv.org/pdf/1502.00956', 'B6_S31_ORB-SLAM_MurArtal_2015.pdf'),
    ('https://arxiv.org/pdf/2003.08934', 'B6_S31_NeRF_Mildenhall_2020.pdf'),
    ('https://arxiv.org/pdf/2308.04079', 'B6_S31_3DGaussianSplatting_Kerbl_2023.pdf'),
    ('https://www.ros.org/reps/rep-0105.html', 'B7_S34_REP-105_marcos_coordenadas.html'),
    ('https://arxiv.org/pdf/1707.06347', 'B8_S39_PPO_Schulman_2017.pdf'),
    ('https://arxiv.org/pdf/1502.05477', 'B8_S39_TRPO_Schulman_2015.pdf'),
    ('https://arxiv.org/pdf/1703.06907', 'B8_S39_DomainRandomization_Tobin_2017.pdf'),
    ('https://arxiv.org/pdf/1910.07113', 'B8_S39_DexterityRubik_OpenAI_2019.pdf'),
    ('https://arxiv.org/pdf/2303.04137', 'B8_S40_DiffusionPolicy_Chi_2023.pdf'),
    ('https://arxiv.org/pdf/2304.13705', 'B8_S40_ACT-ALOHA_Zhao_2023.pdf'),
    ('https://arxiv.org/pdf/2307.15818', 'B8_S40_RT-2_Brohan_2023.pdf'),
    ('https://arxiv.org/pdf/2406.09246', 'B8_S40_OpenVLA_Kim_2024.pdf'),
    ('https://arxiv.org/pdf/2410.24164', 'B8_S40_pi0_Black_2024.pdf'),
    ('https://arxiv.org/pdf/2503.14734', 'B8_S40_GR00T-N1_NVIDIA_2025.pdf'),
    ('https://arxiv.org/pdf/2503.20020', 'B8_S40_GeminiRobotics_DeepMind_2025.pdf'),
    ('http://incompleteideas.net/book/RLbook2020.pdf', 'B8_LIBRO_SuttonBarto_RL_Introduction_2ed_2018.pdf'),
]

ctx = ssl.create_default_context()
ctx.check_hostname = False
ctx.verify_mode = ssl.CERT_NONE
op = urllib.request.build_opener(urllib.request.HTTPSHandler(context=ctx))
op.addheaders = [('User-Agent', 'Mozilla/5.0')]
urllib.request.install_opener(op)

ok = fallo = salt = 0
for url, nombre in FICHEROS:
    ruta = os.path.join(DESTINO, nombre)
    if os.path.exists(ruta) and os.path.getsize(ruta) > 20000:
        print(f'ya estaba   {nombre}'); salt += 1; continue
    try:
        urllib.request.urlretrieve(url, ruta)
        mb = os.path.getsize(ruta) / 1048576
        print(f'descargado  {nombre}  ({mb:.1f} MB)'); ok += 1
    except Exception as e:
        print(f'ERROR       {nombre}: {e}'); fallo += 1

print(f'\nResumen: {ok} descargados, {salt} ya estaban, {fallo} con error.')
print(f'Carpeta: {DESTINO}')
```
